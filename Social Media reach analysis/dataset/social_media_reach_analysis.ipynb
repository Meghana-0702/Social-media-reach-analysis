```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use("ggplot")
```

```python

df = pd.read_csv("Social_media.csv")
```

```python
df.head()
```

```python
df.shape
df.info()
df.describe()
```

```python
print(df.columns)
```

```python
df.isnull().sum()
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["reach"], kde=True)
plt.show()
```

```python
df.sort_values("reach", ascending=False).head(10)
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x="likes", y="reach", data=df)
plt.show()
```

```python
sns.scatterplot(x="comments", y="reach", data=df)
plt.show()
```

```python
sns.scatterplot(x="shares", y="reach", data=df)
plt.show()
```
